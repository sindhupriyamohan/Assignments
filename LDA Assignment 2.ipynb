```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

# -------------------------
# Generate Data
# -------------------------
np.random.seed(42)

n = 200

# Class 0
x0 = np.random.normal(0, 1, n)
y0 = np.random.normal(0, 5, n)

# Class 1
x1 = np.random.normal(3, 1, n)
y1 = np.random.normal(0, 5, n)

# Combine data
X0 = np.column_stack((x0, y0))
X1 = np.column_stack((x1, y1))

X = np.vstack((X0, X1))
y = np.array([0]*n + [1]*n)

# -------------------------
# Plot Original Data
# -------------------------
plt.figure(figsize=(6, 5))

plt.scatter(x0, y0, alpha=0.5, label="Class 0")
plt.scatter(x1, y1, alpha=0.5, label="Class 1")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Original Dataset")
plt.legend()
plt.show()

# -------------------------
# Apply PCA
# -------------------------
pca = PCA(n_components=1)
X_pca = pca.fit_transform(X)

# -------------------------
# Apply LDA
# -------------------------
lda = LinearDiscriminantAnalysis(n_components=1)
X_lda = lda.fit_transform(X, y)

# -------------------------
# PCA Projection
# -------------------------
plt.figure(figsize=(7, 4))

plt.hist(X_pca[y == 0],
         bins=20,
         alpha=0.6,
         label="Class 0")

plt.hist(X_pca[y == 1],
         bins=20,
         alpha=0.6,
         label="Class 1")

plt.title("PCA Projection")
plt.xlabel("Principal Component 1")
plt.legend()
plt.show()

# -------------------------
# LDA Projection
# -------------------------
plt.figure(figsize=(7, 4))

plt.hist(X_lda[y == 0],
         bins=20,
         alpha=0.6,
         label="Class 0")

plt.hist(X_lda[y == 1],
         bins=20,
         alpha=0.6,
         label="Class 1")

plt.title("LDA Projection")
plt.xlabel("Linear Discriminant 1")
plt.legend()
plt.show()
```
